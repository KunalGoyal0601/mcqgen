# MCQ Generator

Generate multiple-choice questions from study notes using a Hugging Face hosted model. Nothing runs until you execute the final cell.

## Setup
1. Install dependencies with python -m pip install -r requirements.txt.
2. Copy .env.example to .env and set HUGGINGFACEHUB_API_TOKEN. Never paste a real token into this notebook or commit it.
3. Run cells in order. Generation requires internet access and model/provider access.


```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
if not (REPO_ROOT / "src" / "mcqgen").is_dir():
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT / "src"))

from mcqgen.mcqgenerater import generate_mcqs
from mcqgen.utils import read_file
```

## Provide source text
Use a local .txt or text-based .pdf file, or paste notes directly. Scanned PDFs may require OCR.


```python
# Option A: read a local file (update the path to your own study material).
# context = read_file(REPO_ROOT / "data" / "study_notes.txt")

# Option B: paste a short excerpt here.
context = """
Photosynthesis is the process by which green plants use sunlight to synthesize food
from carbon dioxide and water. Chlorophyll absorbs light energy. Oxygen is released
as a by-product, and glucose stores chemical energy.
""".strip()
```

## Generate questions
This cell makes a hosted-model API call and may be subject to provider-side usage limits or charges.


```python
questions = generate_mcqs(context, num_questions=3)
for index, item in enumerate(questions, start=1):
    print(f"\n{index}. {item['question']}")
    for option_index, option in enumerate(item["options"], start=1):
        print(f"   {option_index}. {option}")
    print(f"Answer: {item['answer']}\nExplanation: {item['explanation']}")
```
